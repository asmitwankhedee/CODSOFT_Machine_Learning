```python
df['dob'] = pd.to_datetime(df['dob'])
df['trans_date_trans_time'] = pd.to_datetime(df['trans_date_trans_time'])

df['age'] = df['trans_date_trans_time'].dt.year - df['dob'].dt.year - (
    (df['trans_date_trans_time'].dt.month < df['dob'].dt.month) |
    ((df['trans_date_trans_time'].dt.month == df['dob'].dt.month) &
     (df['trans_date_trans_time'].dt.day < df['dob'].dt.day))
)
```

```python
df['hour'] = df['trans_date_trans_time'].dt.hour
```

```python
import numpy as np

df['distance'] = np.sqrt(
    (df['lat'] - df['merch_lat'])**2 +
    (df['long'] - df['merch_long'])**2
)
```

```python
df=df.drop(columns=['dob','trans_date_trans_time'])
```

```python
df
```

```python
features=df[['merchant', 'category', 'amt', 'gender', 'lat', 'long', 'city_pop', 'job', 'unix_time', 'merch_lat', 'merch_long', 'is_fraud','age','hour','distance']]
```

```python
col_features=features.select_dtypes(include='object')
```

```python
from sklearn.preprocessing import LabelEncoder
encoder=LabelEncoder()
```

```python
for col in col_features:
  features[col]=encoder.fit_transform(features[col])
```

```python
features
```

```python
X=features.drop('is_fraud',axis=1)
y=features['is_fraud']
```

```python
from xgboost import XGBClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import confusion_matrix, classification_report
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.20,random_state=42)
```

```python
scaler=StandardScaler()
```

```python
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

```python
model=XGBClassifier()
```

```python
model.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)
```

```python
print(confusion_matrix(y_test,y_pred))
```

```python
print(classification_report(y_test,y_pred))
```
